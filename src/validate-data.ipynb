```python
# ==========================================
# CHECK NULL VALUES IN KAGGLE CSV DATASET
# ==========================================

import os
import pandas as pd

# ------------------------------------------
# 1. Find CSV file
# ------------------------------------------

csv_file = None

for root, dirs, files in os.walk("/kaggle/input"):
    for file in files:
        if file.lower().endswith(".csv"):
            csv_file = os.path.join(root, file)
            break

    if csv_file:
        break


# ------------------------------------------
# 2. Check whether CSV was found
# ------------------------------------------

if csv_file is None:

    print("ERROR: No CSV file found in /kaggle/input")

else:

    print("CSV file found:")
    print(csv_file)


    # --------------------------------------
    # 3. Load CSV file
    # --------------------------------------

    df = pd.read_csv(csv_file)

    print("\nFile loaded successfully!")


    # --------------------------------------
    # 4. Display first 5 rows
    # --------------------------------------

    print("\n==========================================")
    print("FIRST 5 ROWS")
    print("==========================================")

    print(df.head())


    # --------------------------------------
    # 5. Dataset information
    # --------------------------------------

    print("\n==========================================")
    print("DATASET INFORMATION")
    print("==========================================")

    print("Number of rows    :", df.shape[0])
    print("Number of columns :", df.shape[1])


    # --------------------------------------
    # 6. Column names
    # --------------------------------------

    print("\n==========================================")
    print("COLUMN NAMES")
    print("==========================================")

    print(df.columns.tolist())


    # --------------------------------------
    # 7. Check NULL values in every column
    # --------------------------------------

    print("\n==========================================")
    print("NULL VALUES IN EACH COLUMN")
    print("==========================================")

    null_values = df.isnull().sum()

    print(null_values)


    # --------------------------------------
    # 8. Total NULL values
    # --------------------------------------

    total_null = df.isnull().sum().sum()

    print("\n==========================================")
    print("TOTAL NULL VALUES")
    print("==========================================")

    print("Total null values:", total_null)


    # --------------------------------------
    # 9. Check whether dataset contains NULL
    # --------------------------------------

    print("\n==========================================")
    print("NULL VALUE STATUS")
    print("==========================================")

    if df.isnull().values.any():
        print("Dataset contains NULL values.")
    else:
        print("Dataset contains NO NULL values.")


    # --------------------------------------
    # 10. NULL percentage in each column
    # --------------------------------------

    print("\n==========================================")
    print("NULL PERCENTAGE")
    print("==========================================")

    null_percentage = (df.isnull().sum() / len(df)) * 100

    print(null_percentage)


    # --------------------------------------
    # 11. Display only columns having NULL
    # --------------------------------------

    print("\n==========================================")
    print("COLUMNS HAVING NULL VALUES")
    print("==========================================")

    columns_with_null = df.isnull().sum()

    columns_with_null = columns_with_null[
        columns_with_null > 0
    ]

    if len(columns_with_null) > 0:
        print(columns_with_null)
    else:
        print("No columns contain NULL values.")


    # --------------------------------------
    # 12. Display rows containing NULL
    # --------------------------------------

    print("\n==========================================")
    print("ROWS CONTAINING NULL VALUES")
    print("==========================================")

    null_rows = df[df.isnull().any(axis=1)]

    if len(null_rows) > 0:
        print(null_rows)
    else:
        print("No rows contain NULL values.")


    # --------------------------------------
    # 13. Final summary
    # --------------------------------------

    print("\n==========================================")
    print("FINAL SUMMARY")
    print("==========================================")

    print("File           :", csv_file)
    print("Rows           :", df.shape[0])
    print("Columns        :", df.shape[1])
    print("Total NULLs    :", total_null)

    if total_null > 0:
        print("Status         : NULL values found")
    else:
        print("Status         : No NULL values found")
```
